### Practical 1D — Handle missing values

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Karan"],
    "Marks": [85, np.nan, 78, 92, np.nan],
    "Attendance": [90, 85, np.nan, 95, 80]
}

df_missing = pd.DataFrame(data)
print("Missing values before cleaning:")
display(df_missing.isnull().sum())

```

```python
#filling missing value with mean
df_missing["Marks"] = df_missing["Marks"].fillna(df_missing["Marks"].mean())
df_missing["Attendance"] = df_missing["Attendance"].fillna(
    df_missing["Attendance"].mean()
)

```

```python
print("Cleaned dataset:")
display(df_missing)
```

```python
#writting data in a file
df_missing.to_csv("cleaned_students.csv", index=False)
```

Mean imputation replaces missing values with the mean of the available values. It is a simple technique; choose a method appropriate to the dataset and task.

### Practical 3B — Extract quotes, authors, and tags

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://quotes.toscrape.com/"
response = requests.get(url, timeout=15)
response.raise_for_status()
soup = BeautifulSoup(response.text, "html.parser")

quotes_data = []
for quote in soup.find_all("div", class_="quote"):
    quote_text = quote.find("span", class_="text").get_text(strip=True)
    author = quote.find("small", class_="author").get_text(strip=True)
    tags = [tag.get_text(strip=True) for tag in quote.select(".tags .tag")]

    quotes_data.append({
        "Quote": quote_text,
        "Author": author,
        "Tags": ", ".join(tags)
    })

quotes_df = pd.DataFrame(quotes_data)
display(quotes_df.head())
print("Total quotes:", len(quotes_df))
quotes_df.to_csv("quotes_dataset.csv", index=False)
print("Saved quotes_dataset.csv")
```

### Practical 3C — Scrape multiple pages

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

all_quotes = []

for page in range(1, 4):
    url = f"https://quotes.toscrape.com/page/{page}/"
    response = requests.get(url, timeout=15)
    response.raise_for_status()
    soup = BeautifulSoup(response.text, "html.parser")

    for quote in soup.select(".quote"):
        all_quotes.append({
            "Page": page,
            "Quote": quote.select_one(".text").get_text(strip=True),
            "Author": quote.select_one(".author").get_text(strip=True)
        })

    time.sleep(1)  # respectful delay

multiple_quotes_df = pd.DataFrame(all_quotes)
print("Total records:", len(multiple_quotes_df))
display(multiple_quotes_df.head())
multiple_quotes_df.to_csv("multiple_page_quotes.csv", index=False)
```

### Practical 4B — Convert API data to CSV

```python
import requests
import pandas as pd

url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url, timeout=15)
response.raise_for_status()
users = response.json()

records = []
for user in users:
    records.append({
        "ID": user["id"],
        "Name": user["name"],
        "Email": user["email"],
        "City": user["address"]["city"],
        "Company": user["company"]["name"]
    })

users_df = pd.DataFrame(records)
display(users_df.head())
print("Dataset shape:", users_df.shape)
print("Missing values:")
display(users_df.isnull().sum())

users_df.to_csv("api_users.csv", index=False)
print("Saved api_users.csv")
```

### Practical 5A — Create database and table

```python
import sqlite3

connection = sqlite3.connect("college_lab.db")
cursor = connection.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    department TEXT NOT NULL,
    marks REAL
)
""")

connection.commit()
print("Database and table created successfully.")
connection.close()
```

### Practical 5B — Insert student records

```python
import sqlite3

connection = sqlite3.connect("college_lab.db")
cursor = connection.cursor()

students = [
    (101, "Rahul", "CSE", 88),
    (102, "Neha", "IT", 92),
    (103, "Amit", "CSE", 76),
    (104, "Priya", "ECE", 85),
    (105, "Karan", "IT", 79)
]

cursor.executemany("""
INSERT OR REPLACE INTO students
(student_id, name, department, marks)
VALUES (?, ?, ?, ?)
""", students)

connection.commit()
print("Student records inserted successfully.")
connection.close()
```

Parameterized queries use `?` placeholders to keep values separate from SQL text. `INSERT OR REPLACE` replaces a row when a primary-key conflict occurs; ordinary `INSERT` may be preferable when replacement is not intended.

### Practical 5C — Retrieve and filter records

```python
import sqlite3

connection = sqlite3.connect("college_lab.db")
cursor = connection.cursor()

cursor.execute("SELECT * FROM students ORDER BY student_id")
print("All student records:")
for row in cursor.fetchall():
    print(row)

cursor.execute("""
SELECT name, marks
FROM students
WHERE marks > ?
ORDER BY marks DESC
""", (80,))

print("\nStudents scoring above 80:")
for row in cursor.fetchall():
    print(row)

connection.close()
```

### Practical 5D — Update and delete records

```python
import sqlite3

connection = sqlite3.connect("college_lab.db")
cursor = connection.cursor()

cursor.execute(
    "UPDATE students SET marks = ? WHERE student_id = ?",
    (90, 101)
)
connection.commit()
print("Student marks updated.")

cursor.execute(
    "DELETE FROM students WHERE student_id = ?",
    (105,)
)
connection.commit()
print("Student record deleted.")

cursor.execute("SELECT * FROM students ORDER BY student_id")
for row in cursor.fetchall():
    print(row)

connection.close()
```

### Practical 5E — SQL aggregation

```python
import sqlite3

connection = sqlite3.connect("college_lab.db")
cursor = connection.cursor()

cursor.execute("""
SELECT COUNT(*), AVG(marks), MAX(marks), MIN(marks)
FROM students
""")
result = cursor.fetchone()

print("Total students:", result[0])
print("Average marks:", result[1])
print("Highest marks:", result[2])
print("Lowest marks:", result[3])

cursor.execute("""
SELECT department, AVG(marks)
FROM students
GROUP BY department
ORDER BY department
""")

print("\nDepartment-wise average marks:")
for row in cursor.fetchall():
    print(row)

connection.close()
```

### Practical 5F — SQLite with Pandas and visualization

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

connection = sqlite3.connect("college_lab.db")

query = """
SELECT student_id, name, department, marks
FROM students
ORDER BY student_id
"""
students_df = pd.read_sql_query(query, connection)
display(students_df)
students_df.info()

students_df.plot(
    kind="bar", x="name", y="marks",
    legend=False, figsize=(8, 5)
)
plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()

students_df.to_csv("database_students.csv", index=False)
connection.close()
```

```python
import requests
import pandas as pd
import sqlite3

# Step 1: Retrieve API data
url = "https://jsonplaceholder.typicode.com/posts"
response = requests.get(url, timeout=15)
response.raise_for_status()
posts = response.json()

print("API data retrieved successfully.")

# Step 2: Convert JSON to a DataFrame
posts_df = pd.DataFrame(posts)
posts_df = posts_df.rename(columns={
    "userId": "user_id",
    "id": "post_id"
})

display(posts_df.head())

# Step 3: Store records in SQLite
connection = sqlite3.connect("api_database.db")
posts_df.to_sql("posts", connection, if_exists="replace", index=False)
print("Records stored in database.")

# Step 4: Analyze using SQL
query = """
SELECT user_id, COUNT(*) AS total_posts
FROM posts
GROUP BY user_id
ORDER BY user_id
"""
result_df = pd.read_sql_query(query, connection)
display(result_df)

# Step 5: Export summary
result_df.to_csv("posts_analysis.csv", index=False)
print("Analysis exported to posts_analysis.csv")

connection.close()
```
