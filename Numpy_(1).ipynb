```python
#importing library
import numpy as np
```

```python
# one dimensional array
arr1= np.array([10,20,30,40])
print(arr1)
```

```python
#two dimensional array
arr2= np.array([[1,2,3],
[4,5,6]])
print(arr2)
```

```python
#three dimensional array
arr3= np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(arr3.ndim)
print(arr3.size)
print(arr3.dtype)
```

```python
#using np.zeros()
arr4=np.zeros((3,4))
print(arr4)
```

```python
#using np.ones()
arr5=np.ones((2,3))
print(arr5)
```

```python
#using np.arange()
arr6=np.arange(0,20,2)
print(arr6)
```

```python
#using np.linspace()
arr7=np.linspace(0,10,5)
print(arr7)
```

```python
#Traditional Python approach:
numbers= [1,2,3,4,5]
result = []
for x in numbers:
  result.append(x*2)
```

```python
#Numpy vectorized approach:
arr8=np.array([1,2,3,4,5])
result = arr8*2
print(result)
```

```python
#Universal function
arr9=np.array([1,2,3,4,5])
result=np.sqrt(arr9)
print(result)


```

```python
#Binary Universal function
a=np.array([1,2,3])
b=np.array([4,5,6])
result=np.add(a,b)
print(result)
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.subtract(a,b))
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
print(np.multiply(a,b))
print(np.divide(a,b))
print(np.maximum(a,b))
```

```python
#Array operations
a=np.array([1,2,3])
b=np.array([4,5,6])
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a**2)
```

```python
#Comparison and boolean indexing
arr=np.array([10,20,30,40])
print(arr>20)
```

```python
#Aggregation operations
arr=np.array([10,20,30,40,50])
print(arr.sum())
print(arr.mean())
print(arr.max())
print(arr.min())
```

```python
arr2= np.array([[1,2,3],
[4,5,6]])
np.sum(arr2)
np.sum(arr2, axis=0)
np.sum(arr2, axis=1)
```

```python
#Reshaping array
arr=np.arange(1,13)
new_arr=arr.reshape(3,4)
print(new_arr)
```

```python
#Flatting array
arr14=np.array([[1,2],[3,4]])
print(arr14.flatten())
```

```python
#Broadcasting
arr=np.array([[1,2,3],[4,5,6]])
print(arr+10)
```
