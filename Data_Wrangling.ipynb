```python
##Example 1: Student marks by department and semester
import pandas as pd

data = {
    "Department": ["CSE", "CSE", "CSE", "DS", "DS", "DS"],
    "Semester": [1, 2, 3, 1, 2, 3],
    "Average_Marks": [76, 81, 79, 82, 85, 88]
}

df = pd.DataFrame(data)

multi = df.set_index(["Department", "Semester"])

print(multi)
```

```python
##Selecting data from a MultiIndex
# Select all semester records for CSE
print(multi.loc["CSE"])

# Select the average marks for CSE, Semester 2
print(multi.loc[("CSE", 2), "Average_Marks"])

# Convert index levels back into ordinary columns
print(multi.reset_index())
```

```python
##3.1 merge(): Match rows using a common key
students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Asha", "Ravi", "Neha", "Imran"],
    "Department": ["CSE", "DS", "CSE", "DS"]
})

marks = pd.DataFrame({
    "Student_ID": [101, 102, 104, 105],
    "Data_Science_Marks": [86, 91, 78, 88]
})

# Keep only IDs present in both datasets
dat1 = pd.merge(students, marks, on="Student_ID", how="inner")
print(dat1)

# Keep every student; unmatched marks become NaN
left = pd.merge(students, marks, on="Student_ID", how="left")
print(left)

# Keep all IDs from both datasets
outer = pd.merge(students, marks, on="Student_ID", how="outer")
print(outer)
```

```python
##3.2 concat(): Stack datasets
jan = pd.DataFrame({
    "Month": ["Jan", "Jan"],
    "Product": ["Pen", "Notebook"],
    "Units": [120, 80]
})

feb = pd.DataFrame({
    "Month": ["Feb", "Feb"],
    "Product": ["Pen", "Notebook"],
    "Units": [150, 95]
})

sales = pd.concat([jan, feb], ignore_index=True)
print(sales)
```

```python
##3.3 join(): Combine using index labels
student_info = students.set_index("Student_ID")

student_marks = marks.set_index("Student_ID")

joined = student_info.join(student_marks, how="left")

print(joined)
```

```python
##4.1 pivot(): Long format to wide format
sales_long = pd.DataFrame({
    "Product": ["Pen", "Pen", "Notebook", "Notebook"],
    "Month": ["Jan", "Feb", "Jan", "Feb"],
    "Units_Sold": [120, 150, 80, 95]
})

sales_wide = sales_long.pivot(
    index="Product",
    columns="Month",
    values="Units_Sold"
)

print(sales_wide)
```

```python
##4.2 pivot_table(): Summarize duplicate observations
transactions = pd.DataFrame({
    "Product": ["Pen", "Pen", "Pen", "Notebook", "Notebook"],
    "Month": ["Jan", "Jan", "Feb", "Jan", "Feb"],
    "Units": [50, 70, 150, 80, 95]
})

summary = transactions.pivot_table(
    index="Product",
    columns="Month",
    values="Units",
    aggfunc="sum",
    fill_value=0
)

print(summary)
```

```python
##4.3 melt(): Wide format to long format
wide = pd.DataFrame({
    "Student": ["Asha", "Ravi", "Neha"],
    "Quiz_1": [8, 7, 9],
    "Quiz_2": [9, 8, 10],
    "Quiz_3": [7, 9, 9]
})

long = wide.melt(
    id_vars="Student",
    var_name="Quiz",
    value_name="Score"
)

print(long)
```

```python
##4.4 stack() and unstack()
marks_by_semester = pd.DataFrame({
    "Semester_1": [76, 82],
    "Semester_2": [81, 85]
}, index=["CSE", "DS"])

stacked = marks_by_semester.stack()
print(stacked)

# Move the inner index level back into columns
print(stacked.unstack())
```

```python
##5. End-to-End Real-World Example: College Performance Analysis
profiles = pd.DataFrame({
    "Student_ID": [201, 202, 203, 204],
    "Name": ["Kiran", "Meera", "Arjun", "Sara"],
    "Department": ["CSE", "DS", "CSE", "DS"]
})

attendance = pd.DataFrame({
    "Student_ID": [201, 202, 203, 204],
    "Attendance_Percent": [92, 78, 85, 96]
})

results = pd.DataFrame({
    "Student_ID": [201, 202, 203, 204],
    "Final_Marks": [88, 73, 81, 94]
})

# Combine datasets by student ID
analysis_df = profiles.merge(attendance, on="Student_ID", how="left")
analysis_df = analysis_df.merge(results, on="Student_ID", how="left")
print(analysis_df)

# Summarize average attendance and marks by department
dept_summary = analysis_df.groupby("Department")[
    ["Attendance_Percent", "Final_Marks"]
].mean()

print(dept_summary)

# Reshape the summary for presentation
wide_marks = analysis_df.pivot_table(
    index="Department",
    values="Final_Marks",
    aggfunc="mean"
)

print(wide_marks)
```
